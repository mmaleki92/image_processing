```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Plot the circle: 1 = x^2 + y^2
fig, ax = plt.subplots()
theta = np.linspace(0, 2 * np.pi, 300)
x = np.cos(theta)
y = np.sin(theta)
ax.plot(x, y, 'b', label='$x^2 + y^2 = 1$')
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_aspect('equal')
ax.legend()
plt.title("Circle: $x^2 + y^2 = 1$")
plt.grid()
plt.show()

# Plot the paraboloid: f(x, y) = x^2 + y^2 (Bowl shape)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
x = np.linspace(-1, 1, 100)
y = np.linspace(-1, 1, 100)
x, y = np.meshgrid(x, y)
z = x**2 + y**2

ax.plot_surface(x, y, z, cmap='viridis', edgecolor='none')
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("f(x, y)")
ax.set_title("Bowl-shaped Paraboloid: $f(x, y) = x^2 + y^2$")
ax.view_init(elev=30, azim=225)  # Adjust view for better bowl perspective
plt.show()
```
